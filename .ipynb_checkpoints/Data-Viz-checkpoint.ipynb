# Introduction to Python

```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

### Line Plot

```python
x = [1, 2, 3, 4, 5]
y = [10, 12, 8, 14, 7]

plt.plot(x, y, marker='o')
plt.title('Line Chart')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```

### Multiple Line Plot

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 15, 13, 17, 19]
y2 = [12, 11, 14, 15, 18]

plt.plot(x, y1, label='Dataset 1')
plt.plot(x, y2, label='Dataset 2', linestyle='--')

plt.title('Multiple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
```

**Subplot for Multiple Plot**

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Subplot 1: x vs y1
axes[0].plot(x, y1, color='blue', label='Dataset 1')
axes[0].set_title('X vs Y1')
axes[0].set_xlabel('X-axis')
axes[0].set_ylabel('Y1 values', color='blue')
axes[0].tick_params(axis='y', labelcolor='blue')
axes[0].grid(True)

# Subplot 2: x vs y2
axes[1].plot(x, y2, color='red', linestyle='--', label='Dataset 2')
axes[1].set_title('X vs Y2')
axes[1].set_xlabel('X-axis')
axes[1].set_ylabel('Y2 values', color='red')
axes[1].tick_params(axis='y', labelcolor='red')
axes[1].grid(True)

# Adjust layout
fig.suptitle('Line Plots in Separate Subplots', fontsize=14)
fig.tight_layout(rect=[0, 0.03, 1, 0.95])
plt.show()
```

### Area Chart (Stacked Line Plot)

```python
plt.figure(figsize=(15, 6))
plt.stackplot(x, y1, y2, labels=['Series1', 'Series2'], colors=['skyblue', 'lightgreen'])
plt.title('Area Chart')
plt.legend(loc='upper left')
plt.show()
```

### Bar Plot

```python
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 8]

plt.bar(categories, values, color='skyblue')
plt.title('Bar Chart')
plt.xlabel('Category')
plt.ylabel('Value')

# Add value labels on top of each bar
for i, value in enumerate(values):
    plt.text(i, value + 0.2, str(value), ha='center', va='bottom', fontsize=10)

plt.ylim(0, max(values) + 2)  # Add extra space for labels
plt.tight_layout()
plt.show()
```

**Horizontal Bar Plot**

```python
plt.barh(categories, values, color='lightgreen')
plt.title('Horizontal Bar Chart')
plt.xlabel('Value')
plt.ylabel('Category')
plt.show()
```

### Stacked Bar Plot

```python
labels = ['G1', 'G2', 'G3']
men = [20, 35, 30]
women = [25, 32, 34]

x = np.arange(len(labels))
width = 0.5

plt.bar(x, men, width, label='Men')
plt.bar(x, women, width, bottom=men, label='Women')
plt.ylabel('Scores')
plt.title('Stacked Bar Chart')
plt.xticks(x, labels)
plt.legend()
plt.show()
```

### Pie Chart

```python
labels = ['Python', 'Java', 'C++', 'Ruby']
sizes = [40, 30, 20, 10]

plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title('Pie Chart')
plt.axis('equal')  # Equal aspect ratio ensures a circle.
plt.show()
```

### Scatter Plot

```python
x = [1, 2, 3, 4, 5]
y = [5, 4, 6, 7, 5]

plt.scatter(x, y, color='red')
plt.title('Scatter Plot')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True)
plt.show()
```

### Histogram

```python
data = [22, 87, 5, 43, 56, 73, 55, 54, 11, 20, 51, 5, 79, 31, 27]

plt.hist(data, bins=5, color='purple', edgecolor='black')
plt.title('Histogram')
plt.xlabel('Value Range')
plt.ylabel('Frequency')
plt.show()
```

### Box Plot

```python
data = [ [7, 15, 13, 22, 29, 30, 34, 45, 49, 50, 55], [7, 20, 20, 22, 29, 30, 34, 45, 49, 50, 55] ]

plt.boxplot(data, label=['A', 'B'])
plt.title('Box Plot')
plt.ylabel('Values')
plt.legend()
plt.show()
```
