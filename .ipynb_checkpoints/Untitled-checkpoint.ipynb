# Testing Data Visualization with Python

```python
import numpy as mp
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

### Line Plot 

```python
x = [1, 2, 3, 4, 5]
y = [10, 12, 8, 14, 7]

plt.plot(x, y, marker='o')
plt.title('Line Chart')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.grid(True)
plt.show()
```
